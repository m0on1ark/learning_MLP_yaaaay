```python
import numpy as np
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt
```

```python
words = open("surnames.txt").read().splitlines()
letters = sorted(list(set(''.join(words))))
letters = ['.'] + letters
stoi = {s: i for i, s in enumerate(letters)}
itos = {i: s for s, i in stoi.items()}
print(len(stoi))
print(stoi)
```

```python
bigrams = {}
for word in words:
    chars = ['.'] + list(word) + ['.']
    for ch1, ch2 in zip(chars, chars[1:]):
        bigram = (ch1, ch2)
        bigrams[bigram] = bigrams.get(bigram, 0) + 1
print(bigrams)    
    

    
```

```python
N = np.zeros((len(letters), len(letters)), dtype=int)
for word in words:
    chars = ['.'] + list(word) + ['.']
    for ch1, ch2 in zip(chars, chars[1:]):
        ix1 = stoi[ch1] 
        ix2 = stoi[ch2]
        N[ix1, ix2] += 1 
    
print(N)
```

```python
plt.figure(figsize=(16, 16))
plt.imshow(N, cmap='Blues')

for i in range(len(letters)):
    for j in range(len(letters)):
        chstr = itos[i] + itos[j]
        plt.text(j, i, chstr, ha='center', va='bottom', color='gray', fontsize=8)
        plt.text(j, i, N[i, j].item(), ha='center', va='top', color='gray', fontsize=8)

plt.axis('off')
plt.show()
```

```python
P = N / N.sum(axis=1, keepdims=True) 
print(P)
```

```python
import numpy as np

g = np.random.default_rng(42) 

for _ in range(10):
    out = []
    ix = stoi['.']
    while True:
        p = P[ix]
        ix = g.choice(len(letters), p=p)
        if ix == 0:  
            break
        out.append(itos[ix])
    print(''.join(out))
```

```python
log_likelihood = 0.0
n = 0

for word in words:
    chars = ['.'] + list(word) + ['.']
    for ch1, ch2 in zip(chars, chars[1:]):
        ix1 = stoi[ch1]
        ix2 = stoi[ch2]
        prob = P[ix1, ix2]
        logprob = np.log(prob)
        log_likelihood += logprob
        n += 1

nll = -log_likelihood
print(f'{nll=}')
print(f'{nll/n=}')
```

```python
xs1 = []
xs2 = []
ys = []
for word in words:
    chars = ['.' , '.'] + list(word) + ['.']
    for ch1, ch2, ch3 in zip(chars, chars[1:], chars[2:]):
        xs1.append(stoi[ch1])
        xs2.append(stoi[ch2])
        ys.append(stoi[ch3])
xs1 = torch.tensor(xs1)
xs2 = torch.tensor(xs2)
ys = torch.tensor(ys)
```

```python
xenc1 = F.one_hot(xs1,num_classes=33).float()
xenc2 = F.one_hot(xs2,num_classes=33).float()
xenc = torch.cat([xenc1,xenc2], dim=1)
print(xenc)
```

```python
g = torch.Generator().manual_seed(2147483647)
W = torch.randn((len(letters) * 2,len(letters)), generator=g, requires_grad=True)
```

```python
logits = xenc @ W
print(logits.shape)
print(xenc.shape)
```

```python
counts = logits.exp()
print(counts)
```

```python
row_sums = counts.sum(1, keepdim=True)
print(row_sums)
```

```python
probs = counts / row_sums
print(probs)
```

```python
loss = -probs[torch.arange(len(ys)), ys].log().mean()
print(loss)
```

```python
xenc1 = F.one_hot(xs1,num_classes=33).float()
xenc2 = F.one_hot(xs2,num_classes=33).float()
xenc = torch.cat([xenc1,xenc2], dim=1)
for k in range(100):
    # forward pass
    logits = xenc @ W
    counts = logits.exp()
    probs = counts / counts.sum(1, keepdim=True)
    loss = -probs[torch.arange(len(ys)), ys].log().mean()
    print(loss.item())
    
    # backward pass
    W.grad = None
    loss.backward()
    
    
    W.data += -10 * W.grad
```

```python
g = torch.Generator().manual_seed(2147483647)

for _ in range(10):
    out = []
    ix1 = stoi['.']
    ix2 = stoi['.']
    while True:
        xenc1 = F.one_hot(torch.tensor([ix1]), num_classes=33).float()
        xenc2 = F.one_hot(torch.tensor([ix2]), num_classes=33).float()
        xenc = torch.cat([xenc1, xenc2], dim=1)
        logits = xenc @ W
        counts = logits.exp()
        p = counts / counts.sum(1, keepdim=True)
        
        ix3 = torch.multinomial(p, num_samples=1, generator=g).item()
        if ix3 == 0:
            break
        out.append(itos[ix3])
        
        
        ix1 = ix2
        ix2 = ix3
    print(''.join(out))
```

```python
xs1 = []
xs2 = []
xs3 = []
ys = [] 
for word in words:
    chars = ['.', '.' ,'.'] + list(word) + ['.']
    for ch1, ch2, ch3, ch4 in zip(chars, chars[1:], chars[2:], chars[3:]):
        xs1.append(stoi[ch1])
        xs2.append(stoi[ch2])
        xs3.append(stoi[ch3])
        ys.append(stoi[ch4])
xs1 = torch.tensor(xs1)
xs2 = torch.tensor(xs2)
xs3 = torch.tensor(xs3)
ys3 = torch.tensor(ys)
```

```python
emb_dim = 10 
layer1 = 100
g = torch.Generator().manual_seed(2147483647)
C = torch.randn((len(letters), emb_dim), generator=g, requires_grad=True)
emb1 = C[xs1]
emb2 = C[xs2]
emb3 = C[xs3]
emb = torch.cat([emb1, emb2, emb3], dim=1)
W1 = torch.randn((30, layer1), generator = g, requires_grad=True)
b1 = torch.randn(layer1, generator = g, requires_grad=True)
h = torch.tanh(emb @ W1 + b1)
h.shape
```

```python
g = torch.Generator().manual_seed(2147483647)
b2 = torch.randn(len(letters), generator = g, requires_grad=True)
W2 = torch.randn((layer1, len(letters)), generator = g, requires_grad=True)
logits = h @ W2 + b2
counts = logits.exp()
p = counts / counts.sum(1, keepdim=True)
print(p)

```

```python
loss = -probs[torch.arange(len(ys)), ys].log().mean()
print(loss)
```

```python
parameters = [C, W1, b1, W2, b2]
batch_size = 32
lre = torch.linspace(-3, 0, 1000)  
lrs = 10**lre


lossi = []
lrsi = []

for i in range(1000):
    # forward pass
    ix = torch.randint(0, xs1.shape[0], (batch_size,))
    emb = torch.cat([C[xs1[ix]], C[xs2[ix]], C[xs3[ix]]], dim=1)
    h = torch.tanh(emb @ W1 + b1)
    logits = h @ W2 + b2
    counts = logits.exp()
    probs = counts / counts.sum(1, keepdim=True)
    loss = -probs[torch.arange(batch_size), ys[ix]].log().mean()
    print(loss.item())

    # backward pass
    for p in parameters:
        p.grad = None
    
    loss.backward()
    lr = lrs[i]
    for p in parameters:
        p.data += -lr * p.grad
    lrsi.append(lre[i].item())
    lossi.append(loss.item())
```
